
# 使用密钥服务器获取 ROS 密钥
sudo apt update
sudo apt install -y curl gnupg lsb-release

curl -sSL 'http://keyserver.ubuntu.com/pks/lookup?op=get&search=0xC1CF6E31E6BADE8868B172B4F42ED6FBAB17C654' | sudo gpg --dearmor -o /usr/share/keyrings/ros-archive-keyring.gpg
```

然后添加 Jazzy 专用源（使用华为云镜像加速）：

```bash
echo "deb [arch=$(dpkg --print-architecture) signed-by=/usr/share/keyrings/ros-archive-keyring.gpg] https://mirrors.huaweicloud.com/ros2/ubuntu noble main" | sudo tee /etc/apt/sources.list.d/ros2.list > /dev/null
```

**第三步：更新并安装 Jazzy 桌面版**

```bash
sudo apt update
sudo apt install ros-jazzy-desktop
```

**第四步：配置环境并验证**

```bash
echo "source /opt/ros/jazzy/setup.bash" >> ~/.bashrc
source ~/.bashrc
```

测试安装是否成功：
```bash
ros2 run demo_nodes_cpp talker
```

如果这条命令能正常输出 `Publishing` 消息，说明 Jazzy 安装成功了。
